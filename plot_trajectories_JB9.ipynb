# Single-trial & trial-averaged ViRMEn trajectories

Load specified ViRMEn sessions for a mouse, plot single-trial trajectories
per session, then the trial-averaged trajectory across sessions.

Standalone: needs only numpy, scipy, pandas, matplotlib.
Run in VSCode with the Python extension (the Run button), or run the `# %%` cells
in the Python Interactive window to view/zoom the plots.

## Imports

```python
import os
import numpy as np
import pandas as pd
import scipy.io
import matplotlib.pyplot as plt

%matplotlib inline
```

## User configuration

Edit `ANIMAL_ID` to select a different animal, and `SESSIONS` to choose which sessions to plot.

```python
# ----------------------------- USER CONFIG -----------------------------
# Root folder containing per-animal subfolders (e.g. Y:\virmen\JB9)
DATA_ROOT_ROOT = r"Y:\virmen"

# Option A: grouped by animal. This is convenient for editing by animal.
SESSIONS_BY_ANIMAL = {
    "JB9": [
    # Expert sessions (200x20_friction05_floorStripe)
        ("260720", "session_2"),
        ("260721", "session_1"),
        ("260722", "session_1"),
        ("260727", "session_1"),
        ("260804", "session_1"),
        ("260805", "session_1"),
        ("260807", "session_1"),
    ],
    "JB1": [
    # Expert sessions (200x20_friction05_floorStripe)
        ("260612", "session_3"),
        ("260615", "session_1"),
        ("260616", "session_1"),
        ("260617", "session_1"),
        ("260618", "session_1"),
        ("260622", "session_1"),
        ("260623", "session_1"),
        ("260707", "session_1"),
        ("260714", "session_1"),
        ("260715", "session_1"),
        ("260717", "session_1"),
        ("260720", "session_1"),
        ("260721", "session_1"),
        ("260727", "session_1"),
        ("260730", "session_1"),
        ("260803", "session_2"),
        ("260805", "session_1"),
        ("260807", "session_1"),
        ("260810", "session_2"),
        ],
}

# Option B: flat list of (animal, date_folder, session_folder)
# You can instead define SESSIONS directly like this (uncomment to use):
# SESSIONS = [
#     ("JB9", "260701", "session_1"),
#     ("JB10", "260720", "session_2"),
# ]

# Build a flat `SESSIONS` list from `SESSIONS_BY_ANIMAL` if it's provided.
SESSIONS = []
if isinstance(SESSIONS_BY_ANIMAL, dict):
    for animal, sess_list in SESSIONS_BY_ANIMAL.items():
        for date, sess in sess_list:
            SESSIONS.append((animal, date, sess))

EXCLUDE_ITI = True     # drop inter-trial-interval samples from trajectories
N_RESAMPLE  = 100      # points per trial when averaging
# -----------------------------------------------------------------------

# Column layout from mouse_imaging/options.py -> ops['virmen_mat_columns']
VIRMEN_COLUMNS = ['world_id', 'dx', 'dy', 'dh', 'x', 'y', 'h_int',
                  'inITI', 'reward', 'dt', 'lick', 'trial']
```

## Helper functions

```python
def wrap_h(h):
    """Wrap angle to [-pi, pi]. (Replicates functions.wrap_h; not needed for
    x/y trajectories, included to mirror the repo's loading.)"""
    return (np.asarray(h) + np.pi) % (2 * np.pi) - np.pi


def load_vr(session_mat):
    """Replicate session.py:load_vr for a local sessionData.mat (MAT v5)."""
    mat = scipy.io.loadmat(session_mat)
    arr = np.asarray(mat['sessionData']).T          # (n_channels, n_samples) -> (n_samples, n_channels)
    cols = list(VIRMEN_COLUMNS)
    ncols = arr.shape[1]
    if ncols > len(cols):                            # extra user channels sit before final 'trial' col
        extra = [f'user{i}' for i in range(ncols - len(cols))]
        cols = cols[:-1] + extra + cols[-1:]
    elif ncols < len(cols):
        raise ValueError(f"{session_mat}: {ncols} channels < {len(cols)} expected columns")
    df = pd.DataFrame(arr, columns=cols)
    df['h'] = wrap_h(df['h_int']) * -1               # clockwise = positive (repo convention)
    df['t'] = df['dt'].cumsum()
    df['trial'] = df['trial'].astype(int)
    df['inITI'] = df['inITI'] > 0
    return df


def session_path(date, session, animal=None):
    """Build path to a sessionData.mat given date, session, and animal ID.
    If `animal` is None, fall back to the old `DATA_ROOT` variable (back-compat)."""
    if animal is None:
        return os.path.join(DATA_ROOT, date, session, 'sessionData.mat')
    return os.path.join(DATA_ROOT_ROOT, animal, date, session, 'sessionData.mat')


def trial_ids(df):
    ids = df.loc[~df['inITI'], 'trial'] if EXCLUDE_ITI else df['trial']
    return sorted(i for i in ids.unique() if i > 0)


def trial_xy(df, trial):
    sel = df['trial'] == trial
    if EXCLUDE_ITI:
        sel &= ~df['inITI']
    return df.loc[sel, 'x'].to_numpy(), df.loc[sel, 'y'].to_numpy()


def resample_trial(x, y, n=N_RESAMPLE):
    """Reparameterize by normalized cumulative path length -> n points.
    (Alternative for linear tracks: interpolate x onto a fixed y grid.)"""
    if len(x) < 2:
        return None
    d = np.concatenate([[0], np.cumsum(np.hypot(np.diff(x), np.diff(y)))])
    if d[-1] == 0:
        return None
    u = d / d[-1]
    ui = np.linspace(0, 1, n)
    return np.interp(ui, u, x), np.interp(ui, u, y)
```

## Load the sessions

```python
sessions = []
for animal, date, sess in SESSIONS:
    path = session_path(date, sess, animal)
    if not os.path.exists(path):
        print(f"Warning: missing file {path}; skipping")
        continue
    try:
        df = load_vr(path)
    except Exception as e:
        print(f"Error loading {path}: {e}; skipping")
        continue
    sessions.append({'animal': animal, 'label': f'{animal}/{date}/{sess}', 'df': df,
                     'trials': trial_ids(df)})
    print(f"Loaded {path}: {len(df)} samples, {len(sessions[-1]['trials'])} trials")
```

## Figure 1 — single-trial trajectories (one subplot per session)

```python
n = len(sessions)
fig1, axes = plt.subplots(1, n, figsize=(5 * n, 5), squeeze=False)
for ax, s in zip(axes[0], sessions):
    tids = s['trials']
    cmap = plt.cm.viridis(np.linspace(0, 1, max(len(tids), 1)))
    for c, tid in zip(cmap, tids):
        x, y = trial_xy(s['df'], tid)
        if len(x) > 1:
            ax.plot(x, y, color=c, lw=0.7, alpha=0.7)
    ax.set_title(f"{s['label']}  ({len(tids)} trials)")
    ax.set_xlabel('x (virmen units)'); ax.set_ylabel('y (virmen units)')
    ax.set_aspect('equal', 'box')
fig1.suptitle('Single-trial trajectories')
fig1.tight_layout()
```

## Figure 2 — trial-averaged trajectory per session (overlaid)

```python
# Figure 2 — trial-averaged trajectory per session (interactive if available)
# Colors: each animal gets a base color; sessions for that animal use lighter/darker shades.
try:
    get_ipython().run_line_magic('matplotlib', 'widget')
    interactive_backend = 'matplotlib_widget'
except Exception:
    interactive_backend = None

# Build animal order and counts
animals = []
for s in sessions:
    a = s.get('animal', 'unknown')
    if a not in animals:
        animals.append(a)
counts = {a: sum(1 for s in sessions if s.get('animal') == a) for a in animals}

import matplotlib.colors as mcolors
import colorsys

def distinct_base_colors(n, s=0.75, v=0.95):
    # return n visually distinct RGB(A) tuples using HSV hue spacing
    cols = []
    for i in range(n):
        h = i / n
        r, g, b = colorsys.hsv_to_rgb(h, s, v)
        cols.append((r, g, b, 1.0))
    return cols

base_list = distinct_base_colors(max(1, len(animals)))
base_colors = {a: base_list[i] for i, a in enumerate(animals)}

# Prepare per-animal session index counters
from collections import defaultdict
animal_idx = defaultdict(int)

# shading helper: mix with white to lighten (factor in [0,1])
def shade_color(base_rgba, factor):
    # base_rgba in (r,g,b,a); factor 0 -> base, 1 -> white
    r, g, b, a = base_rgba
    r2 = r * (1 - factor) + 1 * factor
    g2 = g * (1 - factor) + 1 * factor
    b2 = b * (1 - factor) + 1 * factor
    return (r2, g2, b2)

if interactive_backend == 'matplotlib_widget':
    fig2, ax = plt.subplots(figsize=(6, 6))
    for s in sessions:
        a = s.get('animal', 'unknown')
        idx = animal_idx[a]
        total = counts.get(a, 1)
        # choose factor between 0 (base) and 0.6 (lighter)
        factor = 0.6 * (idx / max(1, total - 1)) if total > 1 else 0.0
        col = shade_color(base_colors[a], factor)
        stack = [resample_trial(*trial_xy(s['df'], t)) for t in s['trials']]
        stack = [r for r in stack if r is not None]
        animal_idx[a] += 1
        if not stack:
            continue
        xs = np.vstack([r[0] for r in stack]); ys = np.vstack([r[1] for r in stack])
        mx, my = xs.mean(0), ys.mean(0)
        ax.plot(mx, my, color=col, lw=2, label=f"{s['label']} (mean of {len(stack)})")
    ax.set_title('Trial-averaged trajectory')
    ax.set_xlabel('x (virmen units)'); ax.set_ylabel('y (virmen units)')
    ax.set_aspect('equal', 'box'); ax.legend()
    fig2.tight_layout()
else:
    try:
        import plotly.graph_objects as go
        fig = go.Figure()
        # reset counters for consistent shading
        animal_idx = defaultdict(int)
        for s in sessions:
            a = s.get('animal', 'unknown')
            idx = animal_idx[a]
            total = counts.get(a, 1)
            factor = 0.6 * (idx / max(1, total - 1)) if total > 1 else 0.0
            base = base_colors.get(a, (0.2, 0.2, 0.2, 1.0))
            shaded_rgb = shade_color(base, factor)
            # convert to hex for plotly
            hexcol = mcolors.to_hex(shaded_rgb)
            stack = [resample_trial(*trial_xy(s['df'], t)) for t in s['trials']]
            stack = [r for r in stack if r is not None]
            animal_idx[a] += 1
            if not stack:
                continue
            xs = np.vstack([r[0] for r in stack]); ys = np.vstack([r[1] for r in stack])
            mx, my = xs.mean(0), ys.mean(0)
            fig.add_trace(go.Scatter(x=mx, y=my, mode='lines', name=f"{s['label']} (mean of {len(stack)})", line=dict(color=hexcol, width=2)))
        fig.update_layout(title='Trial-averaged trajectory', xaxis_title='x (virmen units)', yaxis_title='y (virmen units)', width=700, height=700)
        fig.show()
    except Exception as e:
        print("Interactive backends not available. Install `ipympl` for %matplotlib widget or `plotly` for interactive figures. Falling back to static matplotlib. Error:", e)
        fig2, ax = plt.subplots(figsize=(6, 6))
        animal_idx = defaultdict(int)
        for s in sessions:
            a = s.get('animal', 'unknown')
            idx = animal_idx[a]
            total = counts.get(a, 1)
            factor = 0.6 * (idx / max(1, total - 1)) if total > 1 else 0.0
            col = shade_color(base_colors[a], factor)
            stack = [resample_trial(*trial_xy(s['df'], t)) for t in s['trials']]
            stack = [r for r in stack if r is not None]
            animal_idx[a] += 1
            if not stack:
                continue
            xs = np.vstack([r[0] for r in stack]); ys = np.vstack([r[1] for r in stack])
            mx, my = xs.mean(0), ys.mean(0)
            ax.plot(mx, my, color=col, lw=2, label=f"{s['label']} (mean of {len(stack)})")
        ax.set_title('Trial-averaged trajectory')
        ax.set_xlabel('x (virmen units)'); ax.set_ylabel('y (virmen units)')
        ax.set_aspect('equal', 'box'); ax.legend()
        fig2.tight_layout()
```
